```python
# Assignment 3 - Data Visualization of 2024 Reportable Communicable Diseases in Toronto 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
```

```python
ID_2024 = pd.read_excel('ID_communical_diseases_2024.xlsx')
ID_2024
```

```python
df_sorted = ID_2024.sort_values("2024 YTD Cases", ascending=False)
df_sorted["Type of Disease"] = df_sorted["Type of Disease"].str.strip()
df_sorted
```

```python

plt.figure(figsize=(12, 8))

# Color mapping
colours = {
    "Sexually Transmitted and Bloodborne": "#8E24AA",
    "Vaccine Preventable": "#FFB300",
    "Outbreaks": "#F4511E",
    "Enteric, Food and Waterborne": "#039BE5",
    "Direct Contact and Respiratory": "#D81B60",
    "Other": "#757575",
    "Vectorborne and Zoonotics": "#43A047"
}

# Create color list
color_list = df_sorted["Type of Disease"].map(colours).fillna("gray")

# Plot bars
plt.bar(df_sorted["Disease"].fillna("Unknown"), df_sorted["2024 YTD Cases"], color=color_list)
plt.xticks(rotation=90, fontsize=10)
plt.xlabel("Disease", fontsize=12)
plt.ylabel("2024 year-to-date Cases", fontsize=12)
plt.title("2024 Reportable Communicable Disease Toronto Cases", fontsize=14)

# Custom legend
legend_elements = [Line2D([0], [0], color=color, lw=4, label=label) for label, color in colours.items()]
plt.legend(handles=legend_elements, title="Type of Disease", fontsize=10, title_fontsize=12, loc='upper right')

plt.tight_layout()
plt.show()
```
